# Python essentials

Small examples of the Python we will use when building AI applications.

Run the code cells from top to bottom.

## Print something

`print()` displays a value below the cell.

```python
print("Hello, Python!")
```

## Variables

A variable gives a value a name so we can use it again.

```python
name = "Sam"
message_count = 3
print(name)
print(message_count)
```

## Common values

Python can work with text, whole numbers, decimal numbers, and true/false values.

```python
topic = "Python"
score = 8
temperature = 21.5
is_ready = True
print(topic, score, temperature, is_ready)
```

## Put values into a message

An f-string inserts a variable where you write `{...}`.

```python
name = "Sam"
print(f"Hello, {name}!")
```

## Clean up text

`.strip()` removes spaces from the ends. `.lower()` makes text lowercase.

```python
text = "  HELLO Python  "
print(text.strip())
print(text.strip().lower())
```

## Make a choice

An `if` statement chooses which code runs. Indented lines belong to that choice.

```python
score = 8
if score >= 7:
    print("This reply looks useful.")
else:
    print("Let us try again.")
```

## Repeat a step

A `for` loop runs the indented code once for each item.

```python
cities = ["Cleveland", "Paris", "Boston"]
for city in cities:
    print(f"Weather for {city}")
```

## Write a function

A function is a reusable set of steps. `return` sends its result back.

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Sam"))
print(greet("Alex"))
```

## Give a function a default

If we leave out `tone`, Python uses `"plain"`.

```python
def build_prompt(task, tone="plain", max_words=120):
    return f"{task} ({tone}, max {max_words} words)"

print(build_prompt("Summarise this"))
print(build_prompt("Summarise this", max_words=60))
print(build_prompt("Summarise this", tone="formal"))
```

## Keep only useful text

Strip each word and keep it only if something remains.

```python
words = ["Paris", "", "Lyon", " "]
clean = []
for word in words:
    if word.strip():
        clean.append(word.strip())
print(clean)
```

## A shorter way to build the same list

This is called a list comprehension. The loop above is fine too.

```python
clean = [word.strip() for word in words if word.strip()]
print(clean)
```

## Put a few steps together

This function cleans a task before putting it into a prompt.

```python
def prepare_prompt(task):
    clean_task = task.strip()
    return build_prompt(clean_task, max_words=60)

print(prepare_prompt("  Explain Python functions  "))
```
