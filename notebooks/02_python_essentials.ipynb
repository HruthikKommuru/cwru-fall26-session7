A function packages steps you can reuse. Defaults supply values when you omit them; named arguments let you change just one setting. An f-string inserts values into text.

```python
def build_prompt(task, tone="plain", max_words=120):
    return f"{task} ({tone}, max {max_words} words)"

print(build_prompt("Summarise this"))
print(build_prompt("Summarise this", max_words=60))
print(build_prompt("Summarise this", tone="formal"))
```

First use a normal loop to remove empty words. `.strip()` removes spaces from both ends; an empty string counts as false in an `if`.

```python
words = ["Paris", "", "Lyon", " "]
clean = []
for w in words:
    if w.strip():
        clean.append(w.strip())
print(clean)
```

A list comprehension is a shorter way to build the same list. Read it as “keep the stripped word for each word whose stripped value is not empty.”

```python
clean = [w.strip() for w in words if w.strip()]
print(clean)
```

**Helper — just run this.** This small pretend model avoids subscriptions and internet calls. You do not need to learn class writing today.

```python
from types import SimpleNamespace
class ChatModel:
    def invoke(self, prompt):
        return SimpleNamespace(content=f"You said: {prompt}")
```

## Using an object

`llm.invoke(...)` calls a function that belongs to `llm`.
`reply.content` reads a value that belongs to `reply`.
The dot means “look inside this object for this name.”

Create a model object, ask it something, then read the reply’s text. This pattern also appears in real model libraries.

```python
llm = ChatModel()
reply = llm.invoke("hi")
print(reply.content)
```
