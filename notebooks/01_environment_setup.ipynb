## Give each project its own toolbox

Project A might need one version of a library while Project B needs another.
A **virtual environment** is a separate folder of Python tools for one project.
Activating it tells your terminal which Python to use.

These are **terminal commands**, shown here as text. Run them from your project
folder, not inside a Python cell. Use one activation command for your system.

```bash
python -m venv .venv
source .venv/bin/activate  # Windows Command Prompt: .venv\Scripts\activate
pip install requests python-dotenv
pip freeze > requirements.txt
```

`pip freeze` records installed versions; it overwrites the named file. Use it in a
scratch project for this exercise, because this workshop's supplied requirements
also include Jupyter tools. On Colab, the hosted Python is already set up.

## A typical small project

```text
my-project/
├── .venv/             # installed tools; never commit this folder
├── .env               # local settings; never commit secrets
├── .gitignore         # files Git should ignore
├── requirements.txt   # libraries needed by this project
├── tools.py           # reusable helper functions
└── main.py            # the program you run
```

For this workshop, copy `.env.example` to `.env` if you want the next cell to
show your name. You can also leave it absent and see the fallback value.

Read settings from `.env`. A fallback keeps the example useful when no name has been set. Only the synthetic demo name is printed.

````python
from dotenv import load_dotenv
load_dotenv()
demo_name = os.getenv("DEMO_NAME", "not set")
print(demo_name)
````
